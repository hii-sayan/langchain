### Google GEMINI INTIGTARION

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
model = init_chat_model("google_genai:gemini-1.0")
res = model.invoke("what is 2 + 2?")
res.content
```

```python
### ChatOPENAI
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-4o-mini")
res = model.invoke("why is the sky blue?")
res.content
```

### Groq model integration

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:llama-3.3-70b-versatile")
res = model.invoke("why do fish swim?")
res
```

```python
import os
from langchain_groq import ChatGroq

model = ChatGroq(model="llama-3.3-70b-versatile")
res = model.invoke("why do dogs bark?")
res.content
```

### STREAM it streams out the output as it gets generated

```python
for chunk in model.stream("why do dogs bark? write me a essay on it 200 words"):
  print(chunk.text, end="|", flush=True)  
```

### Batch it provides inp to the LLM in a parallel way improves cost, performance by runnning in parallel

```python
res = model.batch([
  "why do parrot talk?",
  "why do dogs bark?",
  "what is quantum computing?"
],)

for r in res:
  print(r.content)
```
