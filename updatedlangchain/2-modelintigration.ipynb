### Models Integrations With OPneAI, Goofle GEmini and GROQ

```python
import os 
from dotenv import load_dotenv
load_dotenv()

os.environ.get("OPENAI_API_KEY") = os.getenv("OPENAI_API_KEY")
os.environ.get("GOOGLE_API_KEY") = os.getenv("GOOGLE_API_KEY")
os.environ.get("GROQ_API_KEY") = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("gpt-3.5-turbo")
model
```

```python
res = model.invoke("hello how are you?")
res
```

```python
res.content
```

### Google GEMINI INTIGTARION

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
model = init_chat_model("google_genai:gemini-1.0")
res = model.invoke("what is 2 + 2?")
res.content
```

```python
### ChatOPENAI
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-4o-mini")
res = model.invoke("why is the sky blue?")
res.content
```

### Groq model integration

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("groq:llama-3.3-70b-versatile")
res = model.invoke("why do fish swim?")
res
```
